## Phase 1: Data Cleaning and Preprocessing

Before we can perform meaningful statistical analysis, we need to ensure our data is clean and in the correct format. This phase focuses on handling missing values, correcting erroneous entries, and standardizing data types.

### 1.1 Handle Missing Values

From `df.info()` and `df.describe()`, we observed missing values in `age`, `quantity`, `discount_pct`, `customer_satisfaction`, and `days_to_ship`. For now, we'll impute numerical missing values with the median, which is more robust to outliers than the mean. For categorical columns, we might fill with a mode or a placeholder like 'Unknown', but for numerical, median is a good start.

`order_id`, `order_date`, `customer_id`, `customer_name`, `gender`, `region`, `city`, `product_category`, `product_name`, `unit_price`, `sales_amount`, `profit`, `shipping_cost`, `payment_method`, `return_flag`, and `order_status` also had a few missing values. For simplicity, we'll drop rows where these core identifiers are missing, as they are fundamental to each record.

```python
# Drop rows with missing essential identifiers/data points
df_clean = df.dropna(subset=['order_id', 'order_date', 'customer_id', 'customer_name', 'gender', 'region', 'city', 'product_category', 'product_name', 'unit_price', 'sales_amount', 'profit', 'shipping_cost', 'payment_method', 'return_flag', 'order_status']).copy()

# Explicitly convert 'return_flag' to boolean here, as it's critical for downstream analysis.
# This ensures that 'False' strings are correctly mapped to boolean False.
df_clean['return_flag'] = df_clean['return_flag'].astype(str).str.lower().map({'true': True, 'false': False})
# Fill any potential NaN from mapping (if original had non-'True'/'False' strings) with False
df_clean['return_flag'].fillna(False, inplace=True)

# Impute missing numerical values with the median after cleaning anomalies (to be done later)
# For now, let's keep track of columns with NaNs that we'll impute
# Let's start by addressing the glaring anomalies first, as they can skew medians.
```

### 1.2 Correct Anomalous Numerical Values

We identified several anomalies:
*   `age`: `min` -7.0, `max` 999.0. Ages should typically be positive and within a reasonable human lifespan (e.g., 0-100).
*   `quantity`: `max` 999.0. A single order item quantity of 999 seems like an outlier.
*   `shipping_cost`: `min` -4.29. Shipping cost cannot be negative.
*   `days_to_ship`: `min` -1.0, `max` 100.0. Shipping days cannot be negative. 100 days is very high and might be an outlier.

We will replace these illogical values with `NaN` first, then impute them or handle them appropriately.

```python
# Correcting 'age' anomalies: replace values outside a plausible range (e.g., 0-100) with NaN
df_clean['age'] = np.where(
    (df_clean['age'] < 0) | (df_clean['age'] > 100),
    np.nan,
    df_clean['age']
)

# Correcting 'quantity' anomaly: replace very high quantity (e.g., > 100) with NaN
df_clean['quantity'] = np.where(
    df_clean['quantity'] > 100, # Assuming a reasonable max quantity for a single item
    np.nan,
    df_clean['quantity']
)

# Correcting 'shipping_cost' anomaly: replace negative values with NaN
df_clean['shipping_cost'] = np.where(
    df_clean['shipping_cost'] < 0,
    np.nan,
    df_clean['shipping_cost']
)

# Correcting 'days_to_ship' anomalies: replace negative or excessively high values (e.g., > 30) with NaN
df_clean['days_to_ship'] = np.where(
    (df_clean['days_to_ship'] < 0) | (df_clean['days_to_ship'] > 30),
    np.nan,
    df_clean['days_to_ship']
)

# Now, impute the missing numerical values (including the ones we just created by replacing anomalies) with the median
for col in ['age', 'quantity', 'discount_pct', 'customer_satisfaction', 'days_to_ship', 'shipping_cost']:
    if col in df_clean.columns:
        median_val = df_clean[col].median()
        df_clean[col].fillna(median_val, inplace=True)

print("Missing values after correcting anomalies and imputation:")
display(df_clean.isnull().sum())
```

### 1.3 Standardize Categorical Values and Convert Data Types

*   `gender`: Inconsistent entries ('F', 'm', 'Other'). We need to standardize these to 'Male', 'Female', 'Other'.
*   `order_date`: Currently `object` type. Needs to be converted to datetime objects for time-series analysis.
*   `return_flag`: Currently `object` type ('True', 'False'). Needs to be converted to boolean.

```python
# Standardize 'gender' column
df_clean['gender'] = df_clean['gender'].str.strip().str.lower().replace({
    'f': 'Female',
    'm': 'Male'
}).str.capitalize()

# Convert 'order_date' to datetime, handling mixed formats and coercing errors
df_clean['order_date'] = pd.to_datetime(df_clean['order_date'], errors='coerce', format='mixed')

# The 'return_flag' conversion was moved to cell 'b882184c' to ensure it happens correctly upfront.
# This line is now redundant and removed.

# Check data types after conversions
print("Data types after cleaning and conversions:")
display(df_clean.dtypes)

# Check unique values in gender after standardization
print("Unique gender values after standardization:")
display(df_clean['gender'].unique())

# Check the distribution of the return_flag to ensure it's balanced
print("Distribution of 'return_flag' after conversion:")
# Ensure the full value counts are displayed
print(df_clean['return_flag'].value_counts())
```

### 1.4 Handle Duplicate Rows and Duplicate `order_id`s

## Phase 4: Hypothesis Testing

Hypothesis testing allows us to draw conclusions about a population based on sample data. We formulate a null hypothesis (H0) and an alternative hypothesis (H1), then use statistical tests to determine if there's enough evidence to reject H0.

We will cover:
*   **Independent Samples t-test**: To compare the means of a numerical variable between two independent groups.
*   **One-way ANOVA**: To compare the means of a numerical variable across three or more independent groups.
*   **Chi-Square Test of Independence**: To test for a significant association between two categorical variables.

### 4.1 Independent Samples t-test

**Purpose**: To determine if there is a statistically significant difference between the means of two independent groups for a continuous dependent variable.

**Example**: Is there a significant difference in `sales_amount` between 'Male' and 'Female' customers?

```python
from scipy import stats

# Prepare data for t-test: sales_amount for Male vs Female
group_male = df_clean[df_clean['gender'] == 'Male']['sales_amount'].dropna()
group_female = df_clean[df_clean['gender'] == 'Female']['sales_amount'].dropna()

# Perform independent samples t-test (Welch's t-test, which doesn't assume equal variances)
t_stat, p_val = stats.ttest_ind(group_male, group_female, equal_var=False)

print(f"--- Independent Samples t-test (Sales Amount by Gender) ---")
print(f"Mean Sales Amount (Male): {group_male.mean():.2f}")
print(f"Mean Sales Amount (Female): {group_female.mean():.2f}")
print(f"t-statistic: {t_stat:.3f}")
print(f"p-value: {p_val:.4f}")

if p_val < 0.05:
    print("Conclusion: Reject the null hypothesis. There is a statistically significant difference in sales amount between male and female customers.")
else:
    print("Conclusion: Fail to reject the null hypothesis. There is no statistically significant difference in sales amount between male and female customers.")
```

### 4.2 One-way ANOVA

**Purpose**: To compare the means of a numerical variable across three or more independent groups. It tests if at least one group mean is different from the others.

**Example**: Is there a significant difference in `profit` across different `product_category`?

```python
# Prepare data for ANOVA: profit by product category
product_categories = df_clean['product_category'].dropna().unique()

# Create a list of profit values for each product category
groups = [df_clean[df_clean['product_category'] == category]['profit'].dropna() for category in product_categories]

# Perform one-way ANOVA
f_stat, p_val = stats.f_oneway(*groups)

print(f"\n--- One-way ANOVA (Profit by Product Category) ---")
print(f"F-statistic: {f_stat:.3f}")
print(f"p-value: {p_val:.4f}")

if p_val < 0.05:
    print("Conclusion: Reject the null hypothesis. There is a statistically significant difference in profit across at least one product category.")
else:
    print("Conclusion: Fail to reject the null hypothesis. There is no statistically significant difference in profit across product categories.")
```

### 4.3 Chi-Square Test of Independence

**Purpose**: To determine if there is a statistically significant association between two categorical variables.

**Example**: Is there a relationship between `region` and `return_flag` (i.e., does the return rate vary significantly by region)?

```python
# Create a contingency table for 'region' and 'return_flag'
contingency_table = pd.crosstab(df_clean['region'], df_clean['return_flag'])

# Perform Chi-square test
chi2, p_val, dof, expected = stats.chi2_contingency(contingency_table)

print(f"\n--- Chi-Square Test of Independence (Region vs. Return Flag) ---")
print("Contingency Table:")
display(contingency_table)
print(f"Chi-square statistic: {chi2:.3f}")
print(f"p-value: {p_val:.4f}")
print(f"Degrees of freedom: {dof}")

if p_val < 0.05:
    print("Conclusion: Reject the null hypothesis. There is a statistically significant association between region and return flag.")
else:
    print("Conclusion: Fail to reject the null hypothesis. There is no statistically significant association between region and return flag.")
```

## Phase 5: Regression Analysis + Assumption Checks

Regression analysis is a powerful statistical tool used to model the relationship between a dependent variable and one or more independent variables. It helps us understand how the value of the dependent variable changes when any one of the independent variables is varied, while the other independent variables are held fixed.

We will cover:
*   **Multiple Linear Regression**: To model a continuous dependent variable.
*   **Logistic Regression**: To model a binary categorical dependent variable.
*   **Assumption Checks**: Key for ensuring the validity of our regression models, including multicollinearity (VIF) and residual analysis.

### 5.1 Multiple Linear Regression

**Purpose**: To model a linear relationship between a continuous dependent variable (e.g., `sales_amount`) and multiple continuous or categorical independent variables.

**Example**: Let's try to predict `sales_amount` based on `quantity`, `unit_price`, `discount_pct`, `profit`, `shipping_cost`, `customer_satisfaction`, `age`, and `days_to_ship`. We'll also include categorical variables like `gender` and `region` by converting them into dummy variables.

```python
import statsmodels.api as sm
from statsmodels.stats.outliers_influence import variance_inflation_factor

# Define our dependent variable
y = df_clean['sales_amount']

# Define potential independent variables (numerical and categorical)
predictors_num = ['quantity', 'unit_price', 'discount_pct', 'profit', 'shipping_cost', 'customer_satisfaction', 'age', 'days_to_ship']
predictors_cat = ['gender', 'region', 'product_category'] # Added product_category for more richness

# Create dummy variables for categorical predictors and explicitly convert them to int
X_cat = pd.get_dummies(df_clean[predictors_cat], drop_first=True).astype(int) # Explicitly convert dummies to int

# Combine numerical and dummy variables
X = pd.concat([df_clean[predictors_num], X_cat], axis=1)

# Ensure all column names are strings and valid Python identifiers
X.columns = X.columns.astype(str)
X.columns = [col.replace('[', '_').replace(']', '_').replace('<', '_').replace('>', '_').replace(' ', '_').replace('-', '_').replace('=', '_') for col in X.columns]

# Add a constant to the independent variables for the intercept term
X = sm.add_constant(X)

# Drop rows with NaN values in X or y if any still exist after previous cleaning
model_df = pd.concat([X, y], axis=1).dropna()
X_model = model_df.drop('sales_amount', axis=1)
y_model = model_df['sales_amount']

# Fit the OLS (Ordinary Least Squares) regression model
model = sm.OLS(y_model, X_model).fit()

print("\n--- Multiple Linear Regression Model Summary (Sales Amount) ---")
print(model.summary())

# Explanation of key summary components:
# R-squared: Explains the proportion of variance in the dependent variable explained by the model.
# Adj. R-squared: Adjusted for the number of predictors.
# F-statistic / Prob (F-statistic): Overall significance of the model.
# Coeff (coef): The estimated regression coefficients for each predictor.
# P>|t|: P-value for each coefficient. Helps determine if a predictor is statistically significant.
# Omnibus / Jarque-Bera: Tests for normality of residuals.
# Durbin-Watson: Tests for autocorrelation in residuals (should be close to 2).
# Condition Number: Indicates multicollinearity (high values > 30 can be problematic).
```

### 5.2 Multicollinearity Check (VIF - Variance Inflation Factor)

**Purpose**: Multicollinearity occurs when independent variables in a regression model are correlated with each other. High multicollinearity can make it difficult to interpret the individual coefficients of the independent variables and can lead to unstable and unreliable regression models.

**VIF (Variance Inflation Factor)**: A common metric to detect multicollinearity. A VIF value greater than 5 (or sometimes 10) is often considered indicative of high multicollinearity.

```python
# Calculate VIF for each independent variable
# We need to drop the constant column for VIF calculation first, then add it back if needed
X_vif = X_model.drop(columns='const', errors='ignore')

# Ensure X_vif is explicitly float type before VIF calculation
X_vif = X_vif.astype(float)

vif_data = pd.DataFrame()
vif_data['feature'] = X_vif.columns
vif_data['VIF'] = [variance_inflation_factor(X_vif.values, i) for i in range(len(X_vif.columns))]

print("\n--- Variance Inflation Factor (VIF) ---")
display(vif_data.sort_values(by='VIF', ascending=False))

print("\nInterpretation: High VIF values (typically > 5 or > 10) suggest multicollinearity. \nIf present, consider removing highly correlated predictors or combining them.")
```

### 5.3 Residual Analysis

**Purpose**: Residuals are the differences between the observed values and the values predicted by the regression model. Analyzing residuals helps verify the assumptions of linear regression:
1.  **Linearity**: The relationship between X and y is linear.
2.  **Independence**: Residuals are independent (no autocorrelation).
3.  **Homoscedasticity**: Residuals have constant variance across all levels of X.
4.  **Normality**: Residuals are normally distributed.

**Methods**:
*   **Residuals vs. Fitted Plot**: To check for linearity and homoscedasticity.
*   **Q-Q Plot**: To check for normality of residuals.

```python
import matplotlib.pyplot as plt
import seaborn as sns
import scipy.stats as stats

# Get residuals and fitted values
residuals = model.resid
fitted_values = model.fittedvalues

# Plot 1: Residuals vs. Fitted Values (check linearity and homoscedasticity)
plt.figure(figsize=(10, 6))
sns.scatterplot(x=fitted_values, y=residuals)
plt.axhline(y=0, color='r', linestyle='--')
plt.xlabel('Fitted Values')
plt.ylabel('Residuals')
plt.title('Residuals vs. Fitted Values Plot')
plt.show()

# Interpretation:
# - No clear pattern or funnel shape: Homoscedasticity looks good.
# - Residuals scattered randomly around zero: Linearity assumption holds.

# Plot 2: Q-Q plot of Residuals (check normality)
plt.figure(figsize=(8, 6))
stats.probplot(residuals, dist="norm", plot=plt)
plt.title('Normal Q-Q Plot of Residuals')
plt.show()

# Interpretation:
# - Points closely follow the 45-degree line: Residuals are approximately normally distributed.
# - Deviations from the line, especially at the tails, suggest non-normality.
```

### 5.4 Logistic Regression

**Purpose**: Used when the dependent variable is binary (e.g., `return_flag` - True/False). It models the probability of the dependent variable being one class (e.g., 'True' for return) given the independent variables.

**Example**: Predict `return_flag` based on `sales_amount`, `discount_pct`, `customer_satisfaction`, `shipping_cost`, `product_category`, `region`, and `gender`.

```python
# Define dependent variable (binary, needs to be 0 or 1)
y_logit = df_clean['return_flag'].astype(int) # True becomes 1, False becomes 0

# Define independent variables (numerical and categorical)
predictors_num_logit = ['sales_amount', 'discount_pct', 'customer_satisfaction', 'shipping_cost', 'days_to_ship']

# Reduce the number of categorical predictors to avoid perfect separation for the logistic regression
# Keeping only 'gender' and 'region' for now as they are less likely to cause perfect separation
predictors_cat_logit = ['gender', 'region']

# Create dummy variables for categorical predictors, ensuring they are int type
X_cat_logit = pd.get_dummies(df_clean[predictors_cat_logit], drop_first=True).astype(int)

# Combine numerical and dummy variables
X_logit = pd.concat([df_clean[predictors_num_logit], X_cat_logit], axis=1)

# Add a constant for the intercept term
X_logit = sm.add_constant(X_logit)

# Ensure all column names are strings and valid Python identifiers
X_logit.columns = X_logit.columns.astype(str)
X_logit.columns = [col.replace('[', '_').replace(']', '_').replace('<', '_').replace('>', '_').replace(' ', '_').replace('-', '_').replace('=', '_') for col in X_logit.columns]

# Drop rows with NaN values in X_logit or y_logit if any still exist
model_logit_df = pd.concat([X_logit, y_logit], axis=1).dropna()
X_logit_model = model_logit_df.drop('return_flag', axis=1)
y_logit_model = model_logit_df['return_flag']

# Explicitly convert X_logit_model to float type to ensure compatibility with statsmodels
X_logit_model = X_logit_model.astype(float)

# Fit the Logistic Regression model
# Using a solver and increasing maxiter to help with convergence, though primary issue is separation
logit_model = sm.Logit(y_logit_model, X_logit_model).fit(method='powell', maxiter=1000)

print("\n--- Logistic Regression Model Summary (Return Flag) ---")
print(logit_model.summary())

print("\n--- Odds Ratios ---")
# Coefficients in logistic regression are in log-odds. Convert to odds ratios for easier interpretation.
# Odds Ratio = exp(coefficient). An odds ratio > 1 means the odds of the event (e.g., return) increase with the predictor.
# An odds ratio < 1 means the odds decrease.
print(np.exp(logit_model.params).round(2))

print("\nInterpretation:")
print("- `P>|z|` column indicates the p-value for each coefficient. Values < 0.05 suggest statistical significance.")
print("- Odds Ratios: For every one-unit increase in the predictor, the odds of the dependent variable being 1 (e.g., a return) change by the factor of the odds ratio, holding other predictors constant.")
```

### 6.3 Time Series Decomposition

Time series decomposition separates a time series into three components: trend, seasonality, and residuals (the irregular component). This helps us understand the underlying patterns and noise in the data.

We'll use `seasonal_decompose` from `statsmodels.tsa.seasonal` for this purpose. It supports both additive and multiplicative models. An additive model is appropriate when the magnitude of seasonal fluctuations does not vary with the level of the time series, while a multiplicative model is more suitable when the seasonal variations change proportionally to the level of the series. For sales data, multiplicative is often a good starting point.

```python
from statsmodels.tsa.seasonal import seasonal_decompose

# Perform seasonal decomposition
# We need to determine the 'period' for seasonality.
# For daily sales, common periods could be 7 (weekly) or 365 (yearly).
# Let's start with a weekly period to observe weekly patterns.

# Ensure there are enough data points for the period
if len(daily_sales) < 2 * 7: # Need at least two full cycles for decomposition
    print("Not enough data for seasonal decomposition with a weekly period.")
else:
    # Changed model to 'additive' to handle zero values in daily_sales
    decomposition = seasonal_decompose(daily_sales, model='additive', period=7, extrapolate_trend='freq')

    # Plot the decomposed components
    fig = decomposition.plot()
    fig.set_size_inches(12, 8)
    fig.suptitle('Time Series Decomposition of Daily Sales (Weekly Period)', y=1.02)
    plt.tight_layout(rect=[0, 0.03, 1, 0.98])
    plt.show()

    print("\nDecomposition components:")
    print("Trend (first 5 values):\n", decomposition.trend.head())
    print("Seasonality (first 5 values):\n", decomposition.seasonal.head())
    print("Residuals (first 5 values):\n", decomposition.resid.head())
```

### 6.6 Stationarity Check (Augmented Dickey-Fuller Test)

Before applying forecasting models, it's crucial to ensure our time series is stationary. A stationary series has constant statistical properties over time (mean, variance, autocorrelation). We'll use the Augmented Dickey-Fuller (ADF) test to check for stationarity.

-   **Null Hypothesis (H0)**: The time series has a unit root (non-stationary).
-   **Alternative Hypothesis (H1)**: The time series does not have a unit root (stationary).

```python
from statsmodels.tsa.stattools import adfuller

def adf_test(series):
    result = adfuller(series, autolag='AIC')
    print('ADF Statistic: %f' % result[0])
    print('p-value: %f' % result[1])
    print('Critical Values:')
    for key, value in result[4].items():
        print('\t%s: %.3f' % (key, value))

    if result[1] <= 0.05: # p-value <= 0.05
        print("Conclusion: Reject H0 (Non-Stationary). Time series is stationary.")
    else:
        print("Conclusion: Fail to Reject H0 (Non-Stationary). Time series is NOT stationary.")

print("\n--- ADF Test on Original Daily Sales Data ---")
adf_test(daily_sales.dropna())
```

### 6.7 Differencing to Achieve Stationarity

Since the original daily sales data is likely not stationary (indicated by a high p-value from the ADF test), we need to apply differencing. First-order differencing (subtracting the value at the previous time step) is a common technique to remove trend and make the series stationary. We will then re-run the ADF test on the differenced series.

```python
# Apply first-order differencing
daily_sales_diff = daily_sales.diff().dropna()

print("\nDifferenced Daily Sales Data Head:")
display(daily_sales_diff.head())

print("\n--- ADF Test on Differenced Daily Sales Data ---")
adf_test(daily_sales_diff)
```

### 6.8 Visualize Differenced Series

Let's visualize the differenced daily sales data to see if the trend has been removed and if the series now appears more stationary.

```python
plt.figure(figsize=(15, 7))
plt.plot(daily_sales_diff.index, daily_sales_diff.values, label='Differenced Daily Sales Amount', color='salmon')
plt.title('Differenced Daily Total Sales Amount Over Time')
plt.xlabel('Date')
plt.ylabel('Differenced Sales Amount')
plt.grid(True, linestyle='--', alpha=0.7)
plt.legend()
plt.tight_layout()
plt.show()
```

### 6.9 ACF and PACF Plots for Differenced Series

After differencing, the ACF and PACF plots should help us identify the appropriate orders for the AR (p) and MA (q) components of an ARIMA model. A stationary series will typically show a quicker decay in the ACF.

```python
from statsmodels.graphics.tsaplots import plot_acf, plot_pacf

# Plot ACF of differenced series
plt.figure(figsize=(15, 6))
plot_acf(daily_sales_diff, lags=30, ax=plt.gca())
plt.title('Autocorrelation Function (ACF) of Differenced Daily Sales')
plt.xlabel('Lags')
plt.ylabel('Autocorrelation')
plt.grid(True, linestyle='--', alpha=0.7)
plt.show()

# Plot PACF of differenced series
plt.figure(figsize=(15, 6))
plot_pacf(daily_sales_diff, lags=30, ax=plt.gca())
plt.title('Partial Autocorrelation Function (PACF) of Differenced Daily Sales')
plt.xlabel('Lags')
plt.ylabel('Partial Autocorrelation')
plt.grid(True, linestyle='--', alpha=0.7)
plt.show()

print("Interpretation:")
print("- ACF: Helps determine the order of the MA (q) component. Look for where the ACF cuts off significantly.")
print("- PACF: Helps determine the order of the AR (p) component. Look for where the PACF cuts off significantly.")
```

### 6.10 Fit SARIMA Model

Now that we have examined the ACF and PACF plots of the differenced series, we can proceed to fit a Seasonal Autoregressive Integrated Moving Average (SARIMA) model. A SARIMA model is denoted as SARIMA(p,d,q)(P,D,Q,S), where:
-   (p,d,q) are the non-seasonal components (AR, I, MA orders).
-   (P,D,Q,S) are the seasonal components (Seasonal AR, Seasonal I, Seasonal MA orders, and the seasonal period).

Based on our previous analysis and general time series practices for daily data:
-   `d=1`: We applied first-order differencing.
-   `S=7`: We observed strong weekly seasonality from decomposition and ACF/PACF plots of the original series.

For initial values of p, q, P, Q, we'll make a conservative choice, often starting with 1 or 0 for the first few lags that are significant. Let's try `p=1`, `q=1` and `P=1`, `Q=1`.

```python
from statsmodels.tsa.statespace.sarimax import SARIMAX

# Define the model parameters based on ACF/PACF of differenced series
# Non-seasonal (p,d,q)
order = (1, 1, 1) # p=1 (AR), d=1 (differencing already applied), q=1 (MA)

# Seasonal (P,D,Q,S)
seasonal_order = (1, 0, 1, 7) # P=1 (Seasonal AR), D=0 (no additional seasonal differencing for now), Q=1 (Seasonal MA), S=7 (weekly seasonality)

print(f"Fitting SARIMA model with order={order} and seasonal_order={seasonal_order}...")

# Fit the SARIMA model
# Exogenous variables can be added if we have other predictors (e.g., holidays, promotions)
# For now, we'll just use the time series itself.

# Ensure daily_sales is clean for modeling, drop any remaining NaNs
daily_sales_for_model = daily_sales.dropna()

# Fit the model
sarima_model = SARIMAX(daily_sales_for_model, order=order, seasonal_order=seasonal_order, enforce_stationarity=False, enforce_invertibility=False)
sarima_results = sarima_model.fit(disp=False) # disp=False suppresses convergence messages

print("\n--- SARIMA Model Summary ---")
print(sarima_results.summary())

print("\nInterpretation:")
print("- AIC and BIC: Lower values indicate a better model fit.")
print("- Ljung-Box test (Prob(Q)): If p-value is high (>0.05), residuals are uncorrelated, indicating a good model.")
print("- Jarque-Bera test (Prob(JB)): If p-value is high (>0.05), residuals are normally distributed.")
print("- Coefficients (coef): Parameter estimates for AR, MA, Seasonal AR, Seasonal MA terms. P>|z| indicates significance.")
```
