# World Top Restaurant Dataset

# About this dataset

This dataset compiles information on 50 of the world's top restaurants, offering a comprehensive look at their operational and culinary aspects. Each entry provides details such as the `Restaurant_Name`, `City`, `Country`, `Cuisine_Type`, `Michelin_Stars`, `Average_Price_USD`, the `Chef`'s name, `Established_Year`, `Seating_Capacity`, `Rating`, `Signature_Dish`, `Years_Operating`, `Price_Category`, `Star_Category`, and `Continent`.

From our initial exploration, we've gathered several key insights:

**Dataset Overview:**
*   The dataset contains 50 unique entries, with no missing values or duplicates, ensuring data quality for analysis.
*   It covers restaurants across four continents: Europe (25), North America (11), Asia (10), and South America (4), with Europe having the highest representation.

**Key Distributions and Relationships:**
*   **Michelin Stars & Price:** A significant portion of these top restaurants boast `Three Stars` (22 out of 50), predominantly falling into the 'Luxury' (`28`) or 'Premium' (`12`) `Price_Category`. This suggests a strong correlation between high culinary accolades and premium pricing.
*   **Geographical Concentration:** Cities like Paris, Singapore, Copenhagen, and New York frequently appear, indicating global culinary hubs for high-end dining.
*   **Operational Aspects:** The `Average_Price_USD` for these establishments ranges from \$85 to \$570, with a mean around \$281, highlighting the exclusivity. `Years_Operating` varies widely, from 8 years up to 128 years, showing a mix of long-standing institutions and newer acclaimed establishments. The `Seating_Capacity` generally ranges from 10 to 80, suggesting an intimate dining experience for most.
*   **Cuisine Diversity:** While `French Contemporary` cuisine is notable, there's a rich diversity of `Cuisine_Type` including Mediterranean, Japanese, and various regional contemporary styles, reflecting a global culinary landscape.

Overall, the dataset provides a snapshot into the world of elite gastronomy, showcasing trends in pricing, location, culinary styles, and the impact of critical acclaim like Michelin stars.

# Univariate Analysis

```python
# Getting a list of numerical columns
numerical_cols = df.select_dtypes(include=np.number).columns.tolist()

# Iterating through numerical columns and create histograms
for col in numerical_cols:
    plt.figure(figsize=(10, 6))
    sns.histplot(data=df, x=col, kde=True)
    plt.title(f'Distribution of {col}')
    plt.xlabel(col)
    plt.ylabel('Frequency')
    plt.show()

# Getting a list of categorical columns
categorical_cols = df.select_dtypes(include='object').columns.tolist()

# Iterate through categorical columns and create countplots
for col in categorical_cols:
    plt.figure(figsize=(10, 6))
    sns.countplot(data=df, x=col)
    plt.title(f'Distribution of {col}')
    plt.xlabel(col)
    plt.ylabel('Count')
    plt.xticks(rotation=45, ha='right')
    plt.tight_layout()
    plt.show()

# We can also use value_counts for a quick look at categorical data distribution
for col in categorical_cols:
    print(f"Value counts for {col}:\n{df[col].value_counts()}\n")
```

# Bivariate Analysis

```python
# Numerical-Numerical relationships (Scatter plots)
numerical_cols = df.select_dtypes(include=np.number).columns.tolist()

# Iterate through pairs of numerical columns and create scatter plots
for i in range(len(numerical_cols)):
    for j in range(i + 1, len(numerical_cols)):
        col1 = numerical_cols[i]
        col2 = numerical_cols[j]
        plt.figure(figsize=(10, 6))
        sns.scatterplot(data=df, x=col1, y=col2)
        plt.title(f'Scatter plot of {col1} vs {col2}')
        plt.xlabel(col1)
        plt.ylabel(col2)
        plt.show()

# Categorical-Numerical relationships (Box plots)
categorical_cols = df.select_dtypes(include='object').columns.tolist()
numerical_cols = df.select_dtypes(include=np.number).columns.tolist()

# Iterate through categorical and numerical columns and create box plots
for cat_col in categorical_cols:
    for num_col in numerical_cols:
        plt.figure(figsize=(10, 6))
        sns.boxplot(data=df, x=cat_col, y=num_col)
        plt.title(f'Box plot of {num_col} by {cat_col}')
        plt.xlabel(cat_col)
        plt.ylabel(num_col)
        plt.xticks(rotation=45, ha='right')
        plt.tight_layout()
        plt.show()
```

## Michelin Stars vs. Average Price

### Subtask:
Investigate the relationship between the number of Michelin Stars and the Average Price (USD) of restaurants, including a visualization to show the correlation.


**Reasoning**:
To analyze the relationship between 'Michelin_Stars' and 'Average_Price_USD', I will generate a scatter plot and calculate their Pearson correlation coefficient as instructed.



```python
plt.figure(figsize=(10, 6))
sns.scatterplot(data=df, x='Michelin_Stars', y='Average_Price_USD', hue='Michelin_Stars', palette='viridis', s=100)
plt.title('Michelin Stars vs. Average Price (USD)', fontsize=16)
plt.xlabel('Michelin Stars', fontsize=12)
plt.ylabel('Average Price (USD)', fontsize=12)
plt.xticks([0, 1, 2, 3])
plt.grid(True)
plt.tight_layout()
plt.show()

correlation = df['Michelin_Stars'].corr(df['Average_Price_USD'])
print(f"Pearson correlation between Michelin Stars and Average Price (USD): {correlation:.2f}")
```

## Price Category vs. Michelin Stars

### Subtask:
Examine how different Price Categories relate to the number of Michelin Stars awarded to restaurants.


To examine how different `Price_Categories` relate to the number of `Michelin_Stars` awarded, I will perform two main steps:
1.  **Visualize the distribution**: I will create a grouped bar chart showing the count of restaurants for each `Star_Category` (derived from `Michelin_Stars`) within each `Price_Category`.
2.  **Quantify the relationship**: I will calculate the average `Michelin_Stars` for each `Price_Category` to provide a numerical summary of this relationship.

**Reasoning**:
To visualize the relationship, I will create a grouped bar chart showing the distribution of 'Star_Category' within each 'Price_Category'.



```python
plt.figure(figsize=(12, 7))
sns.countplot(data=df, x='Price_Category', hue='Star_Category', palette='viridis', order=df['Price_Category'].value_counts().index)
plt.title('Distribution of Michelin Star Categories by Price Category', fontsize=16)
plt.xlabel('Price Category', fontsize=12)
plt.ylabel('Number of Restaurants', fontsize=12)
plt.xticks(rotation=45, ha='right')
plt.legend(title='Star Category')
plt.tight_layout()
plt.show()
```

**Reasoning**:
After visualizing the distribution, I will now quantify the relationship by calculating the average Michelin Stars for each Price Category as stated in the plan.



```python
avg_stars_per_price_category = df.groupby('Price_Category')['Michelin_Stars'].mean().sort_values(ascending=False)
print("Average Michelin Stars per Price Category:\n", avg_stars_per_price_category)
```
