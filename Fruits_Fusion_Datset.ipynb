# Fruit Fusion Dataset

# About the Dataset and Insights

This dataset, `Fruit Fusion`, provides a comprehensive look into various characteristics of 39 different fruits. Each fruit is described by 12 attributes, offering insights into its nutritional content, physical properties, and seasonality.

**Numerical Features:**

- **`sugar_g`** and **`calories`**: These features show a wide range, with most fruits having moderate values, but outliers like 'Date' significantly skewing the maximums (66g sugar, 282 calories per 100g), indicating a high-energy fruit. There's a strong positive correlation between sugar and calories.
- **`acidity_pH`**: Ranges from 2.3 to 6.3, with an unusual high value of 282 for 'Date', suggesting a data anomaly or a specific preparation method. Most fruits fall within the typical acidic range (3-5 pH).
- **`avg_weight_g`**: Shows extreme variation, from 1g ('Cranberry') to 10,000g ('Jackfruit'), highlighting the diverse sizes of fruits included.
- **`water_percent`**: Most fruits are high in water (74-92%), except for 'Date' (21%), reflecting its concentrated nature.
- **`fiber_g_per_100g`**: Varies from 0.4g ('Watermelon') to 9.0g ('Date'), indicating different dietary fiber contributions.
- **`shelf_life_days`**: Ranges from 3 to 365 days, with 'Date' having an exceptionally long shelf life.

**Categorical Features:**

- **`season`**: 'Summer' is the most common season for fruits (18 out of 39), followed by 'Fall'. Some fruits are available 'All' seasons.
- **`top_vitamin`** and **`top_mineral`**: Vitamin C (23 fruits) and Potassium (32 fruits) are overwhelmingly the most common top vitamin and mineral, respectively, underscoring their importance in fruit nutrition.
- **`taste_profile`**: The majority of fruits (30) are categorized as 'Sweet', with 'Tart', 'Sour', 'Creamy', and 'Bitter' being less common.

**Key Relationships:**

- Strong correlations exist between nutritional aspects like `sugar_g` and `calories`. `water_percent` tends to be negatively correlated with `sugar_g` and `calories`.
- Categorical features like `season` and `taste_profile` influence nutritional values; for instance, 'Sweet' fruits generally have higher sugar and calories, and 'Summer' fruits often exhibit higher sugar content. Outliers like 'Date' and 'Jackfruit' significantly impact the distribution of several numerical features.

# Univariate Analysis

```python
# Get a list of numerical columns
numerical_cols = df.select_dtypes(include=np.number).columns.tolist()

# Iterate through numerical columns and create histograms
for col in numerical_cols:
    plt.figure(figsize=(10, 6))
    sns.histplot(data=df, x=col, kde=True)
    plt.title(f'Distribution of {col}')
    plt.xlabel(col)
    plt.ylabel('Frequency')
    plt.show()

# Get a list of categorical columns
categorical_cols = df.select_dtypes(include='object').columns.tolist()

# Iterate through categorical columns and create countplots
for col in categorical_cols:
    plt.figure(figsize=(10, 6))
    sns.countplot(data=df, x=col)
    plt.title(f'Distribution of {col}')
    plt.xlabel(col)
    plt.ylabel('Count')
    plt.xticks(rotation=45, ha='right')
    plt.tight_layout()
    plt.show()

# We can also use value_counts for a quick look at categorical data distribution
for col in categorical_cols:
    print(f"Value counts for {col}:\n{df[col].value_counts()}\n")
```

# Bivariate Analysis

```python
# Numerical-Numerical relationships (Scatter plots)
numerical_cols = df.select_dtypes(include=np.number).columns.tolist()

# Iterate through pairs of numerical columns and create scatter plots
for i in range(len(numerical_cols)):
    for j in range(i + 1, len(numerical_cols)):
        col1 = numerical_cols[i]
        col2 = numerical_cols[j]
        plt.figure(figsize=(10, 6))
        sns.scatterplot(data=df, x=col1, y=col2)
        plt.title(f'Scatter plot of {col1} vs {col2}')
        plt.xlabel(col1)
        plt.ylabel(col2)
        plt.show()

# Categorical-Numerical relationships (Box plots)
categorical_cols = df.select_dtypes(include='object').columns.tolist()
numerical_cols = df.select_dtypes(include=np.number).columns.tolist()

# Iterate through categorical and numerical columns and create box plots
for cat_col in categorical_cols:
    for num_col in numerical_cols:
        plt.figure(figsize=(10, 6))
        sns.boxplot(data=df, x=cat_col, y=num_col)
        plt.title(f'Box plot of {num_col} by {cat_col}')
        plt.xlabel(cat_col)
        plt.ylabel(num_col)
        plt.xticks(rotation=45, ha='right')
        plt.tight_layout()
        plt.show()
```

```python
sns.pairplot(df.select_dtypes(include=np.number))
plt.suptitle('Pairplot of Numerical Features', y=1.02)
plt.show()
```

```python
plt.figure(figsize=(15, 8))
sns.heatmap(df.corr(numeric_only = True), annot=True, cmap="coolwarm", linewidths=0.5)
plt.title("Correlation Matrix", fontsize=14)
plt.show()
```
