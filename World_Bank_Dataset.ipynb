# Handling Missing Data

```python
numerical_cols = df.select_dtypes(include=[np.number])
categorical_cols = df.select_dtypes(include=['object'])

# For numerical columns, we might impute with the mean or median
for col in numerical_cols:
    if df[col].isnull().any():
        df[col].fillna(df[col].mean(), inplace=True)

# For categorical columns, we might impute with the mode or a constant
for col in categorical_cols:
    if df[col].isnull().any():
        df[col].fillna(df[col].mode()[0], inplace=True)

print("\nMissing values after handling:")
print(df.isnull().sum())
```

# Handling Outliers

```python
Q1 = df['Tax Revenue (% of GDP)'].quantile(0.25)
Q3 = df['Tax Revenue (% of GDP)'].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

df_outliers = df[(df['Tax Revenue (% of GDP)'] >= lower_bound) & (df['Tax Revenue (% of GDP)'] <= upper_bound)].copy()

print('Before removing outlier:',len(df))
print('After remvoing outliers:',len(df_outliers))
```

# Visualizing Outliers

```python
# Creating a figure with two subplots (1 row, 2 columns)
fig, axes = plt.subplots(1, 2, figsize=(12, 5))

sns.boxplot(data=df, y='Tax Revenue (% of GDP)', ax=axes[0])
axes[0].set_title('Box plot Before Outlier Removal')
axes[0].set_ylabel('Tax Revenue (% of GDP)')

sns.boxplot(data=df_outliers, y='Tax Revenue (% of GDP)', ax=axes[1])
axes[1].set_title('Box plot After Outlier Removal')
axes[1].set_ylabel('Tax Revenue (% of GDP)')

plt.tight_layout()

plt.show()
```

# Univariate Analysis

```python
# Get a list of numerical columns
numerical_cols = df.select_dtypes(include=np.number).columns.tolist()

# Iterate through numerical columns and create histograms
for col in numerical_cols:
    plt.figure(figsize=(10, 6))
    sns.histplot(data=df, x=col, kde=True)
    plt.title(f'Distribution of {col}')
    plt.xlabel(col)
    plt.ylabel('Frequency')
    plt.show()

# Get a list of categorical columns
categorical_cols = df.select_dtypes(include='object').columns.tolist()

# Iterate through categorical columns and create countplots
for col in categorical_cols:
    plt.figure(figsize=(10, 6))
    sns.countplot(data=df, x=col)
    plt.title(f'Distribution of {col}')
    plt.xlabel(col)
    plt.ylabel('Count')
    plt.xticks(rotation=45, ha='right')
    plt.tight_layout()
    plt.show()

# We can also use value_counts for a quick look at categorical data distribution
for col in categorical_cols:
    print(f"Value counts for {col}:\n{df[col].value_counts()}\n")
```

# Bivariate analysis

```python
# Select two numerical columns for a scatter plot
plt.figure(figsize=(10, 6))
# Print columns to verify the exact name
print(df.columns)
# Correct the column name string if it doesn't match the output above
sns.scatterplot(data=df, x='Tax Revenue (% of GDP)', y='Government Expense (% of GDP)')
plt.title('Scatter Plot of Tax Revenue vs. Government Debt')
plt.xlabel('Tax Revenue (% of GDP)')
plt.ylabel('Government Debt (% of GDP)')
plt.show()

# Select one numerical and one categorical column for a box plot
plt.figure(figsize=(12, 6))
sns.boxplot(data=df, x='GDP (Current USD)', y='Government Expense (% of GDP)')
plt.title('Box Plot of Government Debt by Income Group')
plt.xlabel('Income Group')
plt.ylabel('Government Debt (% of GDP)')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()

# Select two categorical columns for a countplot or heatmap
plt.figure(figsize=(12, 6))
sns.countplot(data=df, x='country_id', hue='country_name')
plt.title('Count of Lending Type by Income Group')
plt.xlabel('Income Group')
plt.ylabel('Count')
plt.xticks(rotation=45, ha='right')
plt.legend(title='Lending Type')
plt.tight_layout()
plt.show()
```

```python
# Calculate and display the correlation matrix for numerical columns
numerical_cols = df.select_dtypes(include=np.number)
correlation_matrix = numerical_cols.corr()

plt.figure(figsize=(12, 10))
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', fmt=".2f")
plt.title('Correlation Matrix of Numerical Features')
plt.show()
```
