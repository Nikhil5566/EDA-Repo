# Venezuela Resource Dependency & Economy(1960-2023)

# About the dataset & Insight

Dataset, spanning from 1960 to 2023, paints a vivid picture of Venezuela's economic journey, primarily highlighting its profound reliance on natural resources. At its core, the data underscores how deeply intertwined Venezuela's economic prosperity, or lack thereof, has been with its vast oil reserves.

Our visualizations revealed that oil rents, as a percentage of GDP, have historically been a significant component of the nation's economy. Similarly, fuel exports consistently dominated merchandise exports, often accounting for an overwhelming majority. This singular focus on oil meant that other potential revenue streams, such as ores and metals, played a much smaller role in the export portfolio.

The consequence of this resource dependency is evident in the country's economic volatility. The GDP growth percentage over time showed a rollercoaster of rapid expansion followed by sharp contractions, indicating an economy susceptible to global oil price fluctuations and internal production changes. When oil prices were high, Venezuela experienced periods of growth; when they plummeted, so did the economy.

In essence, the insights reveal a nation whose economic fate has been largely dictated by its natural endowments. While these resources have provided immense wealth, they have also contributed to an undiversified economy, leaving it vulnerable to external shocks and cycles of boom and bust. Understanding these trends is crucial for comprehending Venezuela's past economic challenges and informing future development strategies.

# Visualizing Trends Over Time

```python
plt.figure(figsize=(14, 7))
sns.lineplot(data=df, x='year', y='gdp_current_usd')
plt.title('GDP Current USD Over Time')
plt.xlabel('Year')
plt.ylabel('GDP Current USD')
plt.grid(True)
plt.show()

plt.figure(figsize=(14, 7))
sns.lineplot(data=df, x='year', y='gdp_growth_pct')
plt.title('GDP Growth Percentage Over Time')
plt.xlabel('Year')
plt.ylabel('GDP Growth (%)')
plt.grid(True)
plt.show()
```

```python
plt.figure(figsize=(14, 7))
sns.lineplot(data=df, x='year', y='oil_rents_pct_gdp')
plt.title('Oil Rents as Percentage of GDP Over Time')
plt.xlabel('Year')
plt.ylabel('Oil Rents (% of GDP)')
plt.grid(True)
plt.show()

plt.figure(figsize=(14, 7))
sns.lineplot(data=df, x='year', y='total_natural_resource_rents_pct_gdp')
plt.title('Total Natural Resource Rents as Percentage of GDP Over Time')
plt.xlabel('Year')
plt.ylabel('Total Natural Resource Rents (% of GDP)')
plt.grid(True)
plt.show()
```

```python
plt.figure(figsize=(14, 7))
sns.lineplot(data=df, x='year', y='fuel_exports_pct_merch_exports')
plt.title('Fuel Exports as Percentage of Merchandise Exports Over Time')
plt.xlabel('Year')
plt.ylabel('Fuel Exports (% of Merch. Exports)')
plt.grid(True)
plt.show()

plt.figure(figsize=(14, 7))
sns.lineplot(data=df, x='year', y='ores_and_metals_exports_pct_merch_exports')
plt.title('Ores and Metals Exports as Percentage of Merchandise Exports Over Time')
plt.xlabel('Year')
plt.ylabel('Ores and Metals Exports (% of Merch. Exports)')
plt.grid(True)
plt.show()
```

# Univariate Analysis

```python
# Getting a list of numerical columns
numerical_cols = df.select_dtypes(include=np.number).columns.tolist()

# Iterating through numerical columns and create histograms
for col in numerical_cols:
    plt.figure(figsize=(10, 6))
    sns.histplot(data=df, x=col, kde=True)
    plt.title(f'Distribution of {col}')
    plt.xlabel(col)
    plt.ylabel('Frequency')
    plt.show()

# Getting a list of categorical columns
categorical_cols = df.select_dtypes(include='object').columns.tolist()

# Iterate through categorical columns and create countplots
for col in categorical_cols:
    plt.figure(figsize=(10, 6))
    sns.countplot(data=df, x=col)
    plt.title(f'Distribution of {col}')
    plt.xlabel(col)
    plt.ylabel('Count')
    plt.xticks(rotation=45, ha='right')
    plt.tight_layout()
    plt.show()

# We can also use value_counts for a quick look at categorical data distribution
for col in categorical_cols:
    print(f"Value counts for {col}:\n{df[col].value_counts()}\n")
```

# Bivariate Analysis

```python
# Numerical-Numerical relationships (Scatter plots)
numerical_cols = df.select_dtypes(include=np.number).columns.tolist()

# Iterate through pairs of numerical columns and create scatter plots
for i in range(len(numerical_cols)):
    for j in range(i + 1, len(numerical_cols)):
        col1 = numerical_cols[i]
        col2 = numerical_cols[j]
        plt.figure(figsize=(10, 6))
        sns.scatterplot(data=df, x=col1, y=col2)
        plt.title(f'Scatter plot of {col1} vs {col2}')
        plt.xlabel(col1)
        plt.ylabel(col2)
        plt.show()

# Categorical-Numerical relationships (Box plots)
categorical_cols = df.select_dtypes(include='object').columns.tolist()
numerical_cols = df.select_dtypes(include=np.number).columns.tolist()

# Iterate through categorical and numerical columns and create box plots
for cat_col in categorical_cols:
    for num_col in numerical_cols:
        plt.figure(figsize=(10, 6))
        sns.boxplot(data=df, x=cat_col, y=num_col)
        plt.title(f'Box plot of {num_col} by {cat_col}')
        plt.xlabel(cat_col)
        plt.ylabel(num_col)
        plt.xticks(rotation=45, ha='right')
        plt.tight_layout()
        plt.show()
```

```python
sns.pairplot(df.select_dtypes(include=np.number))
plt.suptitle('Pairplot of Numerical Features', y=1.02)
plt.show()
```
