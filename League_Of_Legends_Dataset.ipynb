# Handling Outliers

```python
Q1 = df['control'].quantile(0.25)
Q3 = df['control'].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

df_outliers = df[(df['control'] >= lower_bound) & (df['control'] <= upper_bound)].copy()

print('Before removing outliers:', len(df))
print('After removing outliers:', len(df_outliers))
```

# Univariate Analysis

```python
# Get a list of numerical columns
numerical_cols = df.select_dtypes(include=np.number).columns.tolist()

# Iterate through numerical columns and create histograms
for col in numerical_cols:
    plt.figure(figsize=(10, 6))
    sns.histplot(data=df, x=col, kde=True)
    plt.title(f'Distribution of {col}')
    plt.xlabel(col)
    plt.ylabel('Frequency')
    plt.show()

# Get a list of categorical columns
categorical_cols = df.select_dtypes(include='object').columns.tolist()

# Iterate through categorical columns and create countplots
for col in categorical_cols:
    plt.figure(figsize=(10, 6))
    sns.countplot(data=df, x=col)
    plt.title(f'Distribution of {col}')
    plt.xlabel(col)
    plt.ylabel('Count')
    plt.xticks(rotation=45, ha='right')
    plt.tight_layout()
    plt.show()

# We can also use value_counts for a quick look at categorical data distribution
for col in categorical_cols:
    print(f"Value counts for {col}:\n{df[col].value_counts()}\n")
```

```python
# 1. Scatter plot
plt.figure(figsize=(10, 6))
sns.scatterplot(data=df_outliers, x='damage', y='difficulty')
plt.title('Scatter plot of Attack vs Defense')
plt.xlabel('Attack')
plt.ylabel('Defense')
plt.show()

# 2. Box plot
plt.figure(figsize=(10, 6))
sns.boxplot(data=df_outliers, x='role', y='adaptivetype')
plt.title('Box plot of Magic by Role')
plt.xlabel('Role')
plt.ylabel('Magic')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()

# 3. Scatter plot
plt.figure(figsize=(10, 6))
sns.scatterplot(data=df_outliers, x='difficulty', y='style')
plt.title('Scatter plot of Difficulty vs Style')
plt.xlabel('Difficulty')
plt.ylabel('Style')
plt.show()

# 4. Box plot
plt.figure(figsize=(10, 6))
sns.boxplot(data=df_outliers, x='role', y='patch')
plt.title('Box plot of Attack by Role')
plt.xlabel('Role')
plt.ylabel('Attack')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()

# 5. Scatter plot
plt.figure(figsize=(10, 6))
sns.scatterplot(data=df_outliers, x='skills', y='skill_r')
plt.title('Scatter plot of Defense vs Magic')
plt.xlabel('Defense')
plt.ylabel('Magic')
plt.show()
```
