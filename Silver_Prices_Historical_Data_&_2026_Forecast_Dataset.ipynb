# Univariate Analysis

```python
# Univariate Analysis: Distributions of Numerical Features
numerical_cols = df.select_dtypes(include=np.number).columns.tolist()

# Calculate dynamic grid size
n_cols = 3
n_rows = (len(numerical_cols) + n_cols - 1) // n_cols

# Plot histograms with KDE in a grid
plt.figure(figsize=(15, 5 * n_rows))
for i, col in enumerate(numerical_cols):
    plt.subplot(n_rows, n_cols, i + 1)
    sns.histplot(data=df, x=col, kde=True)
    plt.title(f'Distribution of {col}')
plt.tight_layout()
plt.show()

# Check for categorical columns
categorical_cols = df.select_dtypes(include='object').columns.tolist()
if not categorical_cols:
    print("\n(No categorical columns found in the dataset)")
else:
    for col in categorical_cols:
        plt.figure(figsize=(10, 6))
        sns.countplot(data=df, x=col)
        plt.title(f'Distribution of {col}')
        plt.show()
```

# Bivariate Analysis

```python
import math

# 1. Identify Numerical Columns
numerical_cols = df.select_dtypes(include=np.number).columns.tolist()

# 2. Generate all pairs (Combinations)
import itertools
pairs = list(itertools.combinations(numerical_cols, 2))
total_plots = len(pairs)

# 3. Setup the Grid (3 columns wide)
cols = 3
rows = math.ceil(total_plots / cols)

# Adjust figure size based on rows
fig, axes = plt.subplots(rows, cols, figsize=(18, 5 * rows))
axes = axes.flatten() # Flatten 2D grid to 1D list for easy looping

# 4. Loop through pairs and plot
for i, (col1, col2) in enumerate(pairs):
    sns.scatterplot(data=df, x=col1, y=col2, ax=axes[i], alpha=0.6, color='teal')
    axes[i].set_title(f'{col1} vs {col2}')

# 5. Hide empty subplots (if any)
for i in range(total_plots, len(axes)):
    fig.delaxes(axes[i])

plt.tight_layout()
plt.show()

# 1. Identify Columns
categorical_cols = df.select_dtypes(include=['object', 'category']).columns.tolist()
numerical_cols = df.select_dtypes(include=np.number).columns.tolist()

# 2. Generate all combinations (Cartesian product)
combinations = [(cat, num) for cat in categorical_cols for num in numerical_cols]
total_plots = len(combinations)

# 3. Setup the Grid (2 columns wide - Boxplots need more width)
cols = 2
rows = math.ceil(total_plots / cols)

if total_plots > 0:
    fig, axes = plt.subplots(rows, cols, figsize=(18, 6 * rows))

    # Handle edge case if there is only 1 plot (axes is not a list)
    if total_plots == 1:
        axes = [axes]
    else:
        axes = axes.flatten()

    # 4. Loop through combinations and plot
    for i, (cat_col, num_col) in enumerate(combinations):
        sns.boxplot(data=df, x=cat_col, y=num_col, ax=axes[i], palette='viridis')
        axes[i].set_title(f'{num_col} by {cat_col}')
        axes[i].tick_params(axis='x', rotation=45) # Rotate x-labels if messy

    # 5. Hide empty subplots
    for i in range(total_plots, len(axes)):
        fig.delaxes(axes[i])

    plt.tight_layout()
    plt.show()
else:
    print("No categorical columns found for box plots.")
```

```python
import scipy.stats as stats
from statsmodels.tsa.stattools import adfuller

print("\n--- Starting Financial Statistical Analysis ---")

# 1. PREPROCESSING: Handle Dates & Calculate Returns
# We cannot analyze prices without time.

if 'Date' in df.columns:
    df['Date'] = pd.to_datetime(df['Date'])
    df = df.sort_values('Date')
    df.set_index('Date', inplace=True)

# Calculate Daily Returns (Percentage Change)
# This is what we analyze statistically, not the raw price.
df['Daily_Return'] = df['Close'].pct_change().dropna()

# 2. STATISTICAL TEST 1: The Augmented Dickey-Fuller Test (Stationarity)
# Null Hypothesis: The Series is Non-Stationary (It has a trend).
# Alternate Hypothesis: The Series is Stationary (It stays around a mean).

def run_adf_test(series, name):
    result = adfuller(series.dropna())
    print(f"\n--- Augmented Dickey-Fuller Test: {name} ---")
    print(f"ADF Statistic: {result[0]:.4f}")
    print(f"P-Value: {result[1]:.4f}")
    if result[1] < 0.05:
        print("✅ Result: Stationary (Stable, good for forecasting)")
    else:
        print("❌ Result: Non-Stationary (Trending, needs differencing)")

# Test Raw Price (Likely Non-Stationary)
run_adf_test(df['Close'], "Raw Silver Price")

# Test Returns (Likely Stationary)
run_adf_test(df['Daily_Return'], "Daily Returns")

# 3. STATISTICAL TEST 2: Normality Test (Shapiro-Wilk)
# Do Silver returns follow a normal Bell Curve?
# Note: We take a sample because Shapiro is sensitive to large datasets
sample_returns = df['Daily_Return'].dropna().sample(min(500, len(df)), random_state=42)
stat, p_val = stats.shapiro(sample_returns)

print(f"\n--- Shapiro-Wilk Test for Normality ---")
print(f"P-Value: {p_val:.10f}")
if p_val > 0.05:
    print("✅ Returns are Normally Distributed.")
else:
    print("❌ Returns are NOT Normal (Likely have 'Fat Tails' or outliers).")

# 4. VISUALIZATION: Returns Distribution & Volatility
fig, axes = plt.subplots(2, 1, figsize=(12, 10))

# Plot A: The Distribution of Returns (The "Bell Curve" check)
sns.histplot(df['Daily_Return'], kde=True, color='blue', bins=50, ax=axes[0])
axes[0].set_title('Distribution of Daily Silver Returns (Risk Analysis)')
axes[0].set_xlabel('Daily % Change')
axes[0].axvline(0, color='red', linestyle='--')

# Plot B: Rolling Volatility (Risk over Time)
# We calculate the Standard Deviation of returns over a 30-day window
df['Volatility_30d'] = df['Daily_Return'].rolling(window=30).std()
df['Volatility_30d'].plot(ax=axes[1], color='orange')
axes[1].set_title('30-Day Rolling Volatility (How "Risky" was Silver?)')
axes[1].set_ylabel('Standard Deviation')
plt.tight_layout()

plt.show()
```

# Decomposition & Autocorrelation

```python
from statsmodels.tsa.seasonal import seasonal_decompose
from statsmodels.graphics.tsaplots import plot_acf, plot_pacf

print("\n--- Final Analysis: Decomposition & Autocorrelation ---")

# 1. PREPARE DATA (Ensure Index is Datetime)
# We use Monthly resampling ('M') for decomposition to make the trend clearer
# than noisy daily data.
df_monthly = df.resample('M')['Close'].mean().dropna()

# 2. DECOMPOSITION (Trend vs Seasonality)
# This splits the data into 3 parts: Trend, Seasonal, Residual
decomposition = seasonal_decompose(df_monthly, model='additive')

fig, axes = plt.subplots(4, 1, figsize=(12, 10))
decomposition.observed.plot(ax=axes[0], color='black')
axes[0].set_title('Observed (Actual Price)')
axes[0].set_ylabel('Price')

decomposition.trend.plot(ax=axes[1], color='blue')
axes[1].set_title('Trend (The Long-Term Direction)')
axes[1].set_ylabel('Trend')

decomposition.seasonal.plot(ax=axes[2], color='green')
axes[2].set_title('Seasonality (Recurring Patterns)')
axes[2].set_ylabel('Seasonal')

decomposition.resid.plot(ax=axes[3], color='red', linestyle='None', marker='o', markersize=3)
axes[3].set_title('Residuals (Random Noise / Unexplained)')
axes[3].set_ylabel('Residual')

plt.tight_layout()
plt.show()

# 3. AUTOCORRELATION (Lag Analysis)
# Does the price 1 month ago predict the price today?
# The blue area is the "Confidence Interval". Bars outside it are significant.
fig, axes = plt.subplots(1, 2, figsize=(16, 5))

plot_acf(df_monthly, lags=50, ax=axes[0])
axes[0].set_title('Autocorrelation (ACF) - Momentum Check')

plot_pacf(df_monthly, lags=50, ax=axes[1])
axes[1].set_title('Partial Autocorrelation (PACF) - Direct Lag Influence')

plt.show()
```

# Feature Engineering

```python
print("\n--- Creating Technical Indicators for ML ---")

# Ensure data is sorted by date
df = df.sort_values('Date')

# 1. LAG FEATURES (The "Memory")
# The model needs to know what happened yesterday to predict today.
df['Lag_1'] = df['Close'].shift(1)  # Price 1 day ago
df['Lag_7'] = df['Close'].shift(7)  # Price 1 week ago (Seasonality)
df['Lag_30'] = df['Close'].shift(30) # Price 1 month ago

# 2. ROLLING WINDOW FEATURES (The "Trend")
# Helps the model see if we are in an uptrend or downtrend.
df['Rolling_Mean_7'] = df['Close'].rolling(window=7).mean()
df['Rolling_Mean_30'] = df['Close'].rolling(window=30).mean()
df['Volatility_7'] = df['Close'].rolling(window=7).std() # Short-term risk

# 3. DATE FEATURES (The "Seasonality")
# Helps the model learn patterns like "Silver goes up in December".
df['Month'] = df.index.month
df['Quarter'] = df.index.quarter
df['DayOfWeek'] = df.index.dayofweek

# 4. TARGET VARIABLE (What we want to predict)
# Option A: Predict the specific Price tomorrow (Regression)
# We shift the Close price BACKWARDS by 1 day.
# Row X's "Target" is the price of Row X+1.
df['Target_Next_Day_Price'] = df['Close'].shift(-1)

# 5. CLEANUP
# Creating lags creates NaN values (e.g., you can't have a 7-day lag for the 1st day).
df_ml = df.dropna().copy()

print("Feature Engineering Complete.")
print(f"Data shape for ML: {df_ml.shape}")
print(df_ml[['Close', 'Lag_1', 'Rolling_Mean_7', 'Target_Next_Day_Price']].tail())

# ==========================================
# STEP 6: TIME-SERIES SPLIT (Crucial!)
# ==========================================
# ERROR WARNING: Never use random shuffle (train_test_split) for Time Series!
# You cannot train on 2024 data to predict 2023. You must slice by time.

train_size = int(len(df_ml) * 0.8)
train = df_ml.iloc[:train_size]
test = df_ml.iloc[train_size:]

print(f"\nTraining on {len(train)} days (Past)")
print(f"Testing on {len(test)} days (Future)")

# Define X (Inputs) and y (Target)
features = ['Lag_1', 'Lag_7', 'Lag_30', 'Rolling_Mean_7', 'Rolling_Mean_30',
            'Volatility_7', 'Month', 'Quarter', 'DayOfWeek']

X_train = train[features]
y_train = train['Target_Next_Day_Price']
X_test = test[features]
y_test = test['Target_Next_Day_Price']

print("Ready for Model Training.")
```

# Machine Learning

```python
import xgboost as xgb
from sklearn.metrics import mean_absolute_error, mean_squared_error
import numpy as np

print("\n--- Starting XGBoost Training ---")

# 1. INITIALIZE & TRAIN MODEL
# Objective: 'reg:squarederror' is standard for predicting prices.
# n_estimators: Number of trees (1000 is robust).
# learning_rate: 0.01 makes it learn slowly but accurately.
reg = xgb.XGBRegressor(
    n_estimators=1000,
    learning_rate=0.01,
    max_depth=5,
    objective='reg:squarederror',
    early_stopping_rounds=50,
    random_state=42
)

# Fit the model (eval_set helps stop training if it stops improving)
reg.fit(
    X_train, y_train,
    eval_set=[(X_train, y_train), (X_test, y_test)],
    verbose=100  # Print progress every 100 trees
)

# 2. MAKE PREDICTIONS
# Predict on the Test Set (The "Future" part of our data)
preds = reg.predict(X_test)

# 3. EVALUATE PERFORMANCE
mae = mean_absolute_error(y_test, preds)
# Fix: Calculate RMSE using np.sqrt() as 'squared=False' is removed in newer sklearn versions
rmse = np.sqrt(mean_squared_error(y_test, preds))

print(f"\n--- Model Performance ---")
print(f"MAE (Average Error): ${mae:.2f}")
print(f"RMSE (Root Mean Sq Error): ${rmse:.2f}")
print(f"Target Price Range: ${y_test.min():.2f} - ${y_test.max():.2f}")

# 4. VISUALIZATION 1: ACTUAL VS PREDICTED
# We create a dataframe to make plotting easier
results_df = y_test.to_frame(name='Actual')
results_df['Predicted'] = preds

plt.figure(figsize=(15, 6))
plt.plot(results_df.index, results_df['Actual'], label='Actual Price', color='blue', alpha=0.6)
plt.plot(results_df.index, results_df['Predicted'], label='AI Prediction', color='orange', alpha=0.8)
plt.title('Silver Price Forecast: AI vs Reality')
plt.xlabel('Date')
plt.ylabel('Price (USD)')
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()

# 5. VISUALIZATION 2: FEATURE IMPORTANCE
# What technical indicator mattered most?
xgb.plot_importance(reg, height=0.5, max_num_features=10, title='What Drives Silver Prices?')
plt.show()
```
