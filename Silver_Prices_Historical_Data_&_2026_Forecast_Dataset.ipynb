# Univariate Analysis

```python
# Univariate Analysis: Distributions of Numerical Features
numerical_cols = df.select_dtypes(include=np.number).columns.tolist()

# Calculate dynamic grid size
n_cols = 3
n_rows = (len(numerical_cols) + n_cols - 1) // n_cols

# Plot histograms with KDE in a grid
plt.figure(figsize=(15, 5 * n_rows))
for i, col in enumerate(numerical_cols):
    plt.subplot(n_rows, n_cols, i + 1)
    sns.histplot(data=df, x=col, kde=True)
    plt.title(f'Distribution of {col}')
plt.tight_layout()
plt.show()

# Check for categorical columns
categorical_cols = df.select_dtypes(include='object').columns.tolist()
if not categorical_cols:
    print("\n(No categorical columns found in the dataset)")
else:
    for col in categorical_cols:
        plt.figure(figsize=(10, 6))
        sns.countplot(data=df, x=col)
        plt.title(f'Distribution of {col}')
        plt.show()
```

```python
# Numerical-Numerical relationships (Scatter plots)
numerical_cols = df.select_dtypes(include=np.number).columns.tolist()

# Iterate through pairs of numerical columns and create scatter plots
for i in range(len(numerical_cols)):
    for j in range(i + 1, len(numerical_cols)):
        col1 = numerical_cols[i]
        col2 = numerical_cols[j]
        plt.figure(figsize=(10, 6))
        sns.scatterplot(data=df, x=col1, y=col2)
        plt.title(f'Scatter plot of {col1} vs {col2}')
        plt.xlabel(col1)
        plt.ylabel(col2)
        plt.show()

# Categorical-Numerical relationships (Box plots)
categorical_cols = df.select_dtypes(include='object').columns.tolist()
numerical_cols = df.select_dtypes(include=np.number).columns.tolist()

# Iterate through categorical and numerical columns and create box plots
for cat_col in categorical_cols:
    for num_col in numerical_cols:
        plt.figure(figsize=(10, 6))
        sns.boxplot(data=df, x=cat_col, y=num_col)
        plt.title(f'Box plot of {num_col} by {cat_col}')
        plt.xlabel(cat_col)
        plt.ylabel(num_col)
        plt.xticks(rotation=45, ha='right')
        plt.tight_layout()
        plt.show()
```

```python
sns.pairplot(df.select_dtypes(include = np.number))
plt.suptitle('Pairplot of Numerical Features', y=1.02)
plt.show()
```

```python
plt.figure(figsize=(15, 8))
sns.heatmap(df.corr(numeric_only = True), annot=True, cmap="coolwarm", linewidths=0.5)
plt.title("Correlation Matrix", fontsize=14)
plt.show()
```

```python
import math

# 1. Identify Numerical Columns
numerical_cols = df.select_dtypes(include=np.number).columns.tolist()

# 2. Generate all pairs (Combinations)
import itertools
pairs = list(itertools.combinations(numerical_cols, 2))
total_plots = len(pairs)

# 3. Setup the Grid (3 columns wide)
cols = 3
rows = math.ceil(total_plots / cols)

# Adjust figure size based on rows
fig, axes = plt.subplots(rows, cols, figsize=(18, 5 * rows))
axes = axes.flatten() # Flatten 2D grid to 1D list for easy looping

# 4. Loop through pairs and plot
for i, (col1, col2) in enumerate(pairs):
    sns.scatterplot(data=df, x=col1, y=col2, ax=axes[i], alpha=0.6, color='teal')
    axes[i].set_title(f'{col1} vs {col2}')

# 5. Hide empty subplots (if any)
for i in range(total_plots, len(axes)):
    fig.delaxes(axes[i])

plt.tight_layout()
plt.show()

# 1. Identify Columns
categorical_cols = df.select_dtypes(include=['object', 'category']).columns.tolist()
numerical_cols = df.select_dtypes(include=np.number).columns.tolist()

# 2. Generate all combinations (Cartesian product)
combinations = [(cat, num) for cat in categorical_cols for num in numerical_cols]
total_plots = len(combinations)

# 3. Setup the Grid (2 columns wide - Boxplots need more width)
cols = 2
rows = math.ceil(total_plots / cols)

if total_plots > 0:
    fig, axes = plt.subplots(rows, cols, figsize=(18, 6 * rows))

    # Handle edge case if there is only 1 plot (axes is not a list)
    if total_plots == 1:
        axes = [axes]
    else:
        axes = axes.flatten()

    # 4. Loop through combinations and plot
    for i, (cat_col, num_col) in enumerate(combinations):
        sns.boxplot(data=df, x=cat_col, y=num_col, ax=axes[i], palette='viridis')
        axes[i].set_title(f'{num_col} by {cat_col}')
        axes[i].tick_params(axis='x', rotation=45) # Rotate x-labels if messy

    # 5. Hide empty subplots
    for i in range(total_plots, len(axes)):
        fig.delaxes(axes[i])

    plt.tight_layout()
    plt.show()
else:
    print("No categorical columns found for box plots.")
```
