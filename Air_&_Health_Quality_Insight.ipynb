```python
import numpy as np
import pandas as pd
```

```python
df = pd.read_csv('air_quality_health_dataset.csv')
df.head()
```

# Basic Aggregations

1. What is the average Air Quality Index (AQI) across all regions in the dataset?

2. How many total hospital visits were recorded in the dataset, and what is the maximum number of emergency visits on any single day?

3. What is the median PM2.5 concentration across all records in the dataset?

4. Calculate the total number of days with school closures across all regions.

5. What is the average temperature and humidity for each region, rounded to two decimal places?
Requires grouping by region and computing multiple aggregations.

```python
print('The average Air Quality Index (AQI) across all regions in the dataset:', df['AQI'].mean())
```

```python
print('Total Hospital Visits:', df['hospital_visits'].sum())
print('Maximum Emergency Visits on any single day:', df['emergency_visits'].max())
```

```python
print('The median PM2.5 concentration across all records in the dataset:', df['PM2.5'].median())
```

```python
print('The total number of days with school closures across all regions:', df['school_closures'].sum())
```

```python
print('The average temperature and humidity for each region:\n', df.groupby('region')[['temperature','humidity']].mean())
```

# Group Comparisons

1. Which region has the highest average AQI, and how does it compare to the region with the lowest average AQI?

2. Compare the average respiratory admissions between days with and without lockdown status. Is there a significant difference?

3. How does the average PM10 concentration differ between regions with high population density (>5000) versus low population density (<5000)?

4. Analyze the average mask usage rate across regions. Which region has the highest mask usage, and how does it compare to the overall average?

5. Is there a significant difference in average hospital visits between days with high wind speed (>10 m/s) and low wind speed (≤10 m/s)?

```python
print('Region with the highest average AQI:', df.groupby('region')['AQI'].mean().idxmax())
print('Region with the lowest average AQI:', df.groupby('region')['AQI'].min().idxmax())
```

```python
print('The average respiratory admissions between days with and without lockdown status: \n', df.groupby('lockdown_status')['respiratory_admissions'].mean())
```

```python
high_density_df = df[df['population_density'] > 5000]
low_density_df = df[df['population_density'] < 5000]

avg_pm10_high_density = high_density_df['PM10'].mean()
avg_pm10_low_density = low_density_df['PM10'].mean()

print(f'Average PM10 in regions with high population density (>5000): {avg_pm10_high_density:.2f}')
print(f'Average PM10 in regions with low population density (<5000): {avg_pm10_low_density:.2f}')
```

```python
print('-> The average mask usage rate across regions. Which region has the highest mask usage: \n', df.groupby('region')['mask_usage_rate'].mean())
print('-> Overall mask usage:', df['mask_usage_rate'].mean().round(2))
```

```python
high_wind = df[df['wind_speed'] > 10]
low_wind = df[df['wind_speed'] < 10]

avg_hosptital_visits_high_wind =  high_wind['hospital_visits'].mean()
avg_hosptital_visits_low_wind =  low_wind['hospital_visits'].mean()

print(f'Average hospital visits in regions with high wind speed (>10 m/s): {avg_hosptital_visits_high_wind:.2f}')
print(f'Average hospital visits in regions with low wind speed (<10 m/s): {avg_hosptital_visits_low_wind:.2f}')
```

# Ranking/Top N

1. Which are the top 5 days with the highest AQI values across all regions, and what were their corresponding regions and PM2.5 levels?

2. Rank the regions by their total number of emergency visits. Which region has the most emergency visits?

3. Identify the top 3 days with the highest O3 (ozone) concentrations in the East region.

4. What are the bottom 5 days in terms of public transport usage across all regions, and what was the lockdown status on those days?

5. Rank the regions by their average green cover percentage. Which region has the lowest green cover, and what is its average AQI?

```python
print('The top 5 days with the highest AQI values across all regions, and what were their corresponding regions and PM2.5 levels: \n', df.nlargest(5, 'AQI')[['date','region','PM2.5']])
```

```python
total_emergency_visits_by_region = df.groupby('region')['emergency_visits'].sum()
region_with_most_emergency_visits = total_emergency_visits_by_region.nlargest(1)

print('The regions by their total number of emergency visits:')
print(total_emergency_visits_by_region.sort_values(ascending=False))
print('\nRegion with the most emergency visits:')
print(region_with_most_emergency_visits)
```

```python
print('The top 3 days with the highest O3 (ozone) concentrations in the East region.:\n', df[df['region'] == 'East'].nlargest(3, 'O3'))
```

```python
print('The regions by their average green cover percentage: \n',df.groupby('region')['green_cover_percentage'].mean())
print('Region which has the lowest green cover ->:', df.groupby('region')['green_cover_percentage'].mean().idxmin())
print('Average AQI in the region:', df[df['region'] == 'Central'].AQI.mean())
```

# Time-based Analysis

1. What is the trend of AQI over the first three months of 2020? Does it increase, decrease, or remain stable?

2. How does the average PM2.5 concentration vary by month across all regions? Provide the results for January, February, and March 2020.

3. Is there a noticeable difference in hospital visits between weekdays and weekends in January 2020?

4. What is the average AQI for each week in February 2020 in the South region?

5. Analyze the change in public transport usage from January to March 2020. Does it correlate with changes in lockdown status?

```python
print('The trend of AQI over the first three months of 2020:', df[df['date'].between('2020-01-01', '2020-03-31')].groupby('date')['AQI'].mean())
```

```python
df['date'] = pd.to_datetime(df['date'])
print('The average PM2.5 concentration vary by month across all regions? Provide the results for January, February, and March 2020.: \n', df[df['date'].between('2020-01-01','2020-03-31')].groupby(df['date'].dt.month)['PM2.5'].mean())
```

```python
print('A noticeable difference in hospital visits between weekdays and weekends in January 2020: \n', df[df['date'].between('2020-01-01', '2020-01-31')].groupby(df['date'].dt.dayofweek)['hospital_visits'].mean())
```

```python
february_2020_south_df = df[(df['date'].dt.year == 2020) & (df['date'].dt.month == 2) & (df['region'] == 'South')]
average_aqi_by_week = february_2020_south_df.groupby(february_2020_south_df['date'].dt.isocalendar().week)['AQI'].mean()

print('Average AQI for each week in February 2020 in the South region:')
print(average_aqi_by_week)
```

```python
jan_march_2020_df = df[df['date'].between('2020-01-01', '2020-03-31')]

# Calculate average public transport usage and lockdown status by month
monthly_trends = jan_march_2020_df.groupby(jan_march_2020_df['date'].dt.month)[['public_transport_usage', 'lockdown_status']].mean()

print('Monthly average public transport usage and lockdown status (January to March 2020):')
print(monthly_trends)

# Optional: Calculate correlation
correlation = monthly_trends['public_transport_usage'].corr(monthly_trends['lockdown_status'])
print(f'\nCorrelation between monthly average public transport usage and lockdown status: {correlation:.2f}')
```
