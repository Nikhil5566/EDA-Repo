```python
import numpy as np
import pandas as pd
```

```python
df = pd.read_csv('job_market_unemployment_trends.csv')
df.head()
```

# Basic Aggregation

1. What is the overall average unemployment rate across all locations in the dataset as of July 29, 2025? How does this compare to the total number of job postings recorded?

2. Calculate the total number of job postings and the overall average college degree percentage across all entries. What might this indicate about the workforce's educational background?

3. What is the overall median unemployment rate across all locations? How does it differ from the mean, and what might this suggest about the data distribution?

```python
# Calculate total job postings and average college degree percentage
print('Total number of job postings recorded: ', df['job_postings'].sum())
print('The overall average unemployment rate across all locations in the dataset', df['college_degree_percentage'].mean())
```

```python
print('The total number of job posted:', df['job_postings'].sum())
print('The overall average college degree percentage across all entries:', df['college_degree_percentage'].mean())
```

```python
# Calculate median and mean unemployment rate
print('The overall mean unemployment rate across all locations:', df['unemployment_rate'].mean())
print('The overall median unemployment rate across all locations:',df['unemployment_rate'].median())
```

# Group Aggregation

1. Compare the average unemployment rate between locations with an average age above 40 and those below 40. Which group shows a higher rate, and what could be a potential reason?

2. Analyze the average number of job postings for locations where the college degree percentage is above 70% versus those below 70%. What trends can you identify regarding education and job opportunities?

3. Compare the average unemployment rate in 2024 to that in 2025 (up to July 29, 2025). Which year has a lower rate, and what might this suggest about recent job market trends?

```python
print('The average unemployment rate between locations with an average age above 40 and those below 40:', df[df['average_age'] > 40].groupby('location')['unemployment_rate'].mean())
print('The average unemployment rate between locations with an average age above 40 :', df[df['average_age'] > 40].groupby('location')['unemployment_rate'].mean())
```

```python
print('The average number of job postings for locations where the college degree percentage above 70% :', df[df['college_degree_percentage'] > 70].groupby('location')['job_postings'].mean())
print('The average number of job postings for locations where the college degree percentage below 70% :', df[df['college_degree_percentage'] <= 70].groupby('location')['job_postings'].mean())
```

```python
df['date'] = pd.to_datetime(df['date'])

# Filter by year
data_2024 = df[df['date'].dt.year == 2024]
data_2025 = df[df['date'].dt.year == 2025]

avg_unemployment_2024 = data_2024['unemployment_rate'].mean()
avg_unemployment_2025 = data_2025['unemployment_rate'].mean()

print(f"Average unemployment rate in 2024: {avg_unemployment_2024:.2f}%")
print(f"Average unemployment rate in 2025 (up to July 29): {avg_unemployment_2025:.2f}%")
print(f"Insight: 2025 ({avg_unemployment_2025:.2f}%) shows a lower rate than 2024 ({avg_unemployment_2024:.2f}%), possibly indicating an improving job market as of mid-2025.")
```

# Ranking / Top-N

1. Identify the top 5 locations with the highest average unemployment rate. What are their average job postings, and how might this guide local employment policies?

2. Rank the locations by the total number of job postings and list the top 3. What are the most frequent in-demand skills among these locations?

3. Find the top 5 locations with the highest average college degree percentage. How do their average ages compare to the overall average age across all locations?

```python
print('The top 5 locations with the highest average unemployment rate:', df.groupby('location')['unemployment_rate'].mean().nlargest(5, ))
print('The top 5 locations with the highest average unemployment rate. What are their average job postings:', df.groupby('location')['job_postings'].mean().nlargest(5, ))
```

```python
print('The locations by the total number of job postings and list the top 3:', df.groupby('location')['job_postings'].sum().nlargest(3))
skill = df.groupby('location')['in_demand_skills'].value_counts()
most_frequent_skill = skill.nlargest(5)
display(most_frequent_skill)
```
