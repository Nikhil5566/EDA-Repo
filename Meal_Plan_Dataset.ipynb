# Univariate Analysis

```python
# Get a list of numerical columns
numerical_cols = df.select_dtypes(include=np.number).columns.tolist()

# Iterate through numerical columns and create histograms
for col in numerical_cols:
    plt.figure(figsize=(10, 6))
    sns.histplot(data=df, x=col, kde=True)
    plt.title(f'Distribution of {col}')
    plt.xlabel(col)
    plt.ylabel('Frequency')
    plt.show()

# Get a list of categorical columns
categorical_cols = df.select_dtypes(include='object').columns.tolist()

# Iterate through categorical columns and create countplots
for col in categorical_cols:
    plt.figure(figsize=(10, 6))
    sns.countplot(data=df, x=col)
    plt.title(f'Distribution of {col}')
    plt.xlabel(col)
    plt.ylabel('Count')
    plt.xticks(rotation=45, ha='right')
    plt.tight_layout()
    plt.show()

# We can also use value_counts for a quick look at categorical data distribution
for col in categorical_cols:
    print(f"Value counts for {col}:\n{df[col].value_counts()}\n")
```

# Bivariate Analysis - Categorical Features

```python
# Exercise Schedule vs Gender
plt.figure(figsize=(12, 6))
sns.countplot(data=df, x='Exercise Schedule', hue='Gender')
plt.title('Exercise Schedule Distribution by Gender')
plt.xlabel('Exercise Schedule')
plt.ylabel('Count')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()

# Meal Plan vs Gender
plt.figure(figsize=(12, 6))
sns.countplot(data=df, x='Meal Plan', hue='Gender')
plt.title('Meal Plan Distribution by Gender')
plt.xlabel('Meal Plan')
plt.ylabel('Count')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```

```python
# Exercise Schedule vs Goal
plt.figure(figsize=(12, 6))
sns.countplot(data=df, x='Exercise Schedule', hue='Goal')
plt.title('Exercise Schedule Distribution by Goal')
plt.xlabel('Exercise Schedule')
plt.ylabel('Count')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()

# Meal Plan vs Goal
plt.figure(figsize=(12, 6))
sns.countplot(data=df, x='Meal Plan', hue='Goal')
plt.title('Meal Plan Distribution by Goal')
plt.xlabel('Meal Plan')
plt.ylabel('Count')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```

```python
# Exercise Schedule vs BMI Category
plt.figure(figsize=(12, 6))
sns.countplot(data=df, x='Exercise Schedule', hue='BMI Category')
plt.title('Exercise Schedule Distribution by BMI Category')
plt.xlabel('Exercise Schedule')
plt.ylabel('Count')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()

# Meal Plan vs BMI Category
plt.figure(figsize=(12, 6))
sns.countplot(data=df, x='Meal Plan', hue='BMI Category')
plt.title('Meal Plan Distribution by BMI Category')
plt.xlabel('Meal Plan')
plt.ylabel('Count')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```

```python
# A contingency table of Gender and Goal
contingency_table = pd.crosstab(df['Gender'], df['Goal'])

# Creating a heatmap of the contingency table
plt.figure(figsize=(8, 6))
sns.heatmap(contingency_table, annot=True, fmt='d', cmap='Blues')
plt.title('Relationship between Gender and Goal')
plt.xlabel('Goal')
plt.ylabel('Gender')
plt.show()
```
