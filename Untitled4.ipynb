# Cow Mastitis(From milk)

# About the dataset and insights

Dataset offers a comprehensive look into various milk parameters and health indicators, specifically designed to help in the detection and understanding of bovine mastitis. It contains 800 observations, each representing data from a specific cow on a particular day, across 9 different features.

Key features include Cow_ID for tracking individual animals, Day to denote the day of observation, and crucial milk quality metrics such as Milk_Temperature, Milk_pH, and Milk_Conductivity. These physiological measurements are fundamental indicators of a cow's overall health and the quality of her milk.

Two particularly significant features for mastitis detection are Somatic_Cell_Count (SCC), a universally accepted biomarker for udder inflammation, and Milk_Yield, which measures the amount of milk produced. The dataset also provides a Clotting column, a binary indicator (0 or 1) that likely signifies abnormal milk coagulation, a common symptom of mastitis. Ultimately, the class1 column serves as our target variable, a binary classification (0 for healthy, 1 for mastitis) that directly indicates the presence of the disease.

Our initial data exploration has shown a very clean dataset, free of missing values or duplicates, which is excellent for reliable analysis. Univariate analysis, through histograms, gives us a clear picture of the distribution of each feature, such as the range of SCC or milk pH levels. Bivariate analysis, using scatter plots and box plots, begins to unveil potential relationships between these variables. For instance, we anticipate that higher Somatic_Cell_Count, elevated Milk_Temperature or Milk_Conductivity, and the presence of Clotting will show strong correlations with a class1 diagnosis (mastitis), while Milk_Yield might decrease. These observed relationships, further emphasized in the correlation heatmap, highlight the dataset’s potential for building effective predictive models for early mastitis detection and improved dairy farm management.

# Univariate Analysis

```python
# Getting a list of numerical columns
numerical_cols = df.select_dtypes(include=np.number).columns.tolist()

# Iterating through numerical columns and create histograms
for col in numerical_cols:
    plt.figure(figsize=(10, 6))
    sns.histplot(data=df, x=col, kde=True)
    plt.title(f'Distribution of {col}')
    plt.xlabel(col)
    plt.ylabel('Frequency')
    plt.show()

# Getting a list of categorical columns
categorical_cols = df.select_dtypes(include='object').columns.tolist()

# Iterate through categorical columns and create countplots
for col in categorical_cols:
    plt.figure(figsize=(10, 6))
    sns.countplot(data=df, x=col)
    plt.title(f'Distribution of {col}')
    plt.xlabel(col)
    plt.ylabel('Count')
    plt.xticks(rotation=45, ha='right')
    plt.tight_layout()
    plt.show()

# We can also use value_counts for a quick look at categorical data distribution
for col in categorical_cols:
    print(f"Value counts for {col}:\n{df[col].value_counts()}\n")
```

# Bivariate Analysis

```python
# Numerical-Numerical relationships (Scatter plots)
numerical_cols = df.select_dtypes(include=np.number).columns.tolist()

# Iterate through pairs of numerical columns and create scatter plots
for i in range(len(numerical_cols)):
    for j in range(i + 1, len(numerical_cols)):
        col1 = numerical_cols[i]
        col2 = numerical_cols[j]
        plt.figure(figsize=(10, 6))
        sns.scatterplot(data=df, x=col1, y=col2)
        plt.title(f'Scatter plot of {col1} vs {col2}')
        plt.xlabel(col1)
        plt.ylabel(col2)
        plt.show()

# Categorical-Numerical relationships (Box plots)
categorical_cols = df.select_dtypes(include='object').columns.tolist()
numerical_cols = df.select_dtypes(include=np.number).columns.tolist()

# Iterate through categorical and numerical columns and create box plots
for cat_col in categorical_cols:
    for num_col in numerical_cols:
        plt.figure(figsize=(10, 6))
        sns.boxplot(data=df, x=cat_col, y=num_col)
        plt.title(f'Box plot of {num_col} by {cat_col}')
        plt.xlabel(cat_col)
        plt.ylabel(num_col)
        plt.xticks(rotation=45, ha='right')
        plt.tight_layout()
        plt.show()
```

```python
sns.pairplot(df.select_dtypes(include = np.number))
plt.suptitle('Pairplot of Numerical Feature', y=1.02)
plt.show()
```

```python
plt.figure(figsize=(15, 8))
sns.heatmap(df.corr(numeric_only = True), annot=True, cmap="coolwarm", linewidths=0.5)
plt.title("Correlation Matrix", fontsize=14)
plt.show()
```
