# Handling Missing Data

```python
numerical_cols = df.select_dtypes(include=np.number).columns
categorical_cols = df.select_dtypes(exclude=np.number).columns

# For numerical columns
for col in numerical_cols:
    if df[col].isnull().any():
        df[col].fillna(df[col].mean(), inplace=True)

# For categorical columns
for col in categorical_cols:
    if df[col].isnull().any():
        df[col].fillna(df[col].mode()[0], inplace=True)

print("\nMissing values after handling:")
print(df.isnull().sum())
```

# Handling Outliers

```python
Q1 = df['Number of keys'].quantile(0.25)
Q3 = df['Number of keys'].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

df_outliers = df[(df['Number of keys'] >= lower_bound) & (df['Number of keys'] <= upper_bound)].copy()

print('Before removing outliers:', len(df))
print('After removing outliers:', len(df_outliers))
```

# Visualizing Outliers

```python
fig, axes = plt.subplots(1, 2, figsize=(15, 6))

sns.boxplot(data=df, y='Number of keys', ax=axes[0])
axes[0].set_title('Box plot Before Outlier Removal')
axes[0].set_ylabel('Number of keys')

sns.boxplot(data=df_outliers, y='Number of keys', ax=axes[1])
axes[1].set_title('Box plot After Outlier Removal')
axes[1].set_ylabel('Number of keys')

plt.tight_layout()

plt.show()
```

# Visualization

```python
plt.figure(figsize=(10, 6))
sns.scatterplot(data=df.sample(n=10000), x='Brand', y='Model', alpha=0.5)
plt.title('Scatter Plot with Sampling and Opacity')
plt.xlabel('column1')
plt.ylabel('column2')
plt.show()
```

```python
df_numeric_dates = df.copy()
df_numeric_dates['Released Year'] = pd.to_numeric(df_numeric_dates['Released Year'], errors='coerce')
df_numeric_dates['Released Announced'] = pd.to_numeric(df_numeric_dates['Released Announced'], errors='coerce')

df_numeric_dates.dropna(subset=['Released Year', 'Released Announced'], inplace=True)

plt.figure(figsize=(10, 6))
plt.hexbin(df_numeric_dates['Released Year'], df_numeric_dates['Released Announced'], gridsize=50, cmap='Blues')
plt.colorbar(label='Count')
plt.title('Hexbin Plot')
plt.xlabel('Released Year')
plt.ylabel('Released Announced')
plt.show()
```

```python
plt.figure(figsize=(12, 6))
sns.boxplot(x='Hardware Designer', y='Hard Disk Capacity', data=df.sample(n=10000))
plt.title('Box Plot')
plt.xlabel('Category')
plt.ylabel('Numerical Value')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
numeric_df = df.select_dtypes(include=np.number)
correlation_matrix = numeric_df.corr()

plt.figure(figsize=(12, 10))
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', fmt=".2f")
plt.title('Correlation Heatmap')
plt.show()
```
