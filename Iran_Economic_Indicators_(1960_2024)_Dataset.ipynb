# About the Dataset and Insights

## Iran Economic Indicators (1960-2024): A Snapshot of Decades

This dataset provides a comprehensive look into the economic and demographic landscape of Iran spanning over six decades, from 1960 to 2024. It's a time-series collection, offering 65 annual records that paint a picture of the nation's development and challenges. Each row represents a specific year, detailing various crucial metrics that economic analysts and policymakers often use to assess a country's health and progress.

The dataset comprises twelve key indicators, including fundamental economic measures like Gross Domestic Product (GDP) in current US dollars, GDP growth rate, and GDP per capita. It also includes vital social and financial statistics such as the total population, population growth rate, and the inflation rate. Furthermore, the data sheds light on international trade dynamics through export and import values, trade as a percentage of GDP, and the official exchange rate. Lastly, it provides insight into the structural composition of the economy with the industry value added as a percentage of GDP. This rich collection allows for a deep dive into Iran's economic evolution, identifying periods of prosperity, downturns, and shifting economic priorities.

### Key Insights from the Data:

*   **Long-term GDP Growth:** Over the period of 1960 to 2024, Iran's GDP in current US dollars shows a general upward trend, indicating significant economic expansion. However, the line plot for GDP reveals noticeable fluctuations, with periods of rapid growth interspersed with sharp declines, often reflecting geopolitical events or shifts in global oil markets.
*   **Volatile Inflation:** The inflation rate in Iran has experienced considerable volatility. While there were years with relatively low inflation, the dataset highlights multiple periods of very high inflation, some even approaching 50%. This suggests persistent challenges in maintaining price stability.
*   **Steady Population Expansion:** Iran's total population has shown consistent and significant growth throughout the decades. The population has steadily increased from approximately 21.4 million in 1960 to over 91.5 million in 2024, indicating a major demographic shift that has implications for labor, consumption, and resource allocation.
*   **Trade Dynamics:** The dataset includes export and import values, as well as trade as a percentage of GDP, which allows for analysis of Iran's engagement with the global economy. These figures likely correlate with GDP performance, showing how trade has contributed to or been affected by the broader economic conditions.
*   **Industry's Role:** The 'Industry_Value_Added_Percent_GDP' column offers a glimpse into the structural changes within the Iranian economy, indicating the relative contribution of the industrial sector over time. Trends here can reflect diversification efforts or reliance on specific industries.
*   **Missing Data Points:** A few columns, such as 'GDP_Growth_Percent', 'Official_Exchange_Rate', and 'Population_Growth_Percent', have isolated missing values. For instance, the 'Official_Exchange_Rate' for 2024 is listed as `NaN`, suggesting the data for the most recent year might be incomplete or not yet finalized.

This dataset serves as an excellent foundation for further detailed analysis into the specific drivers behind these trends, and for forecasting future economic trajectories.

# Visualizing the trends of key economic indicators over time using line plots.

```python
plt.figure(figsize=(14, 7))
sns.lineplot(data=df, x='Year', y='GDP_Current_USD')
plt.title('GDP (Current USD) Over Time')
plt.xlabel('Year')
plt.ylabel('GDP (Current USD)')
plt.grid(True)
plt.show()
```

```python
plt.figure(figsize=(14, 7))
sns.lineplot(data=df, x='Year', y='Inflation_Rate_Percent')
plt.title('Inflation Rate (%) Over Time')
plt.xlabel('Year')
plt.ylabel('Inflation Rate (%)')
plt.grid(True)
plt.show()
```

```python
plt.figure(figsize=(14, 7))
sns.lineplot(data=df, x='Year', y='Population_Total')
plt.title('Total Population Over Time')
plt.xlabel('Year')
plt.ylabel('Total Population')
plt.grid(True)
plt.show()
```

# Univariate Analysis

```python
# Getting a list of numerical columns:
numerical_cols = df.select_dtypes(include = np.number).columns.tolist()

# Iteratting through numerical columns and creating histogram:
for col in numerical_cols:
  plt.figure(figsize  =(10, 6))
  sns.histplot(data = df, x = col, kde = True)
  plt.title(f'Distribution of {col}')
  plt.xlabel(col)
  plt.ylabel('Frequency')
  plt.show()

# Getting a list of categorical columns:
categorical_cols = df.select_dtypes(include = 'object').columns.tolist()

# Iteratting through categorical columns and creating countplot
for col in categorical_cols:
  plt.figure(figsize = (10, 6))
  sns.countplot(data = df, x = col)
  plt.title(f'Distribution of{col}')
  plt.xlabel(col)
  plt.ylabel('Count')
  plt.xticks(rotation = 45, ha = 'right')
  plt.tight_layout()
  plt.show()

# We can also use value_count for a quick look at categorical data distribution
for col in categorical_cols:
  print(f'Value counts for {col}: \n {df[col].value_count()}\n')
```

# Bivariate Analysis

```python
# Numerical-Numerical relationships (Scatter plots)
numerical_cols = df.select_dtypes(include=np.number).columns.tolist()

# Iterate through pairs of numerical columns and create scatter plots
for i in range(len(numerical_cols)):
    for j in range(i + 1, len(numerical_cols)):
        col1 = numerical_cols[i]
        col2 = numerical_cols[j]
        plt.figure(figsize=(10, 6))
        sns.scatterplot(data=df, x=col1, y=col2)
        plt.title(f'Scatter plot of {col1} vs {col2}')
        plt.xlabel(col1)
        plt.ylabel(col2)
        plt.show()

# Categorical-Numerical relationships (Box plots)
categorical_cols = df.select_dtypes(include='object').columns.tolist()
numerical_cols = df.select_dtypes(include=np.number).columns.tolist()

# Iterate through categorical and numerical columns and create box plots
for cat_col in categorical_cols:
    for num_col in numerical_cols:
        plt.figure(figsize=(10, 6))
        sns.boxplot(data=df, x=cat_col, y=num_col)
        plt.title(f'Box plot of {num_col} by {cat_col}')
        plt.xlabel(cat_col)
        plt.ylabel(num_col)
        plt.xticks(rotation=45, ha='right')
        plt.tight_layout()
        plt.show()
```

```python
sns.pairplot(df.select_dtypes(include = np.number))
plt.suptitle('Pairplot of Numerical Feature', y = 1.02)
plt.show()
```
